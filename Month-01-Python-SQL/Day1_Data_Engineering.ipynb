```python
import pandas as pd

#Extract: Read the CSV file
df = pd.read_csv("day1_employee_data.csv")

#Transform: Calculate annual salary
df["AnnualSalary"] = df["MonthlySalary"] * 12

#Display
print(df)
```

```python
#To save the transform data
df.to_csv("cleaned_employees.csv", index=False)
```

```python
#Exercise1 : Display only the employees in the IT department

#Select employees working in IT
it_employees = df[df["Department"] == "IT"]

#Display
print(it_employees)
```

```python

#Exercise 2: Display only employees in Finance
#Select employees working in Finance
finance_employee = df[df["Department"] == "Finance"]
 #Display
print(finance_employee)
```

```python
#Exercise 3: To calculate the average monthly salary

#Step1: Extract csv file
df= pd.read_csv("cleaned_employees.csv")

#Transform: Calculate the average monthly salary
average_salary = df["MonthlySalary"].mean()

#Display
print("Average Monthly Salary:", average_salary)
```

```python
#Exercise 4: Find the employee with the highest salary

# find the row with the highest salary
highest_salary = df.loc[df["MonthlySalary"].idxmax()]

#Display
print(highest_salary)
```
